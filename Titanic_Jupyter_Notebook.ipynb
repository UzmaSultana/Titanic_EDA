```python
categorical_cols = Titanic.select_dtypes(include=['object']).columns

# Compute correlation matrix for all columns (numeric + encoded categorical)
corr = Titanic.corr()

# Create the heatmap
sns.heatmap(corr, annot=True, cmap='coolwarm')

# Add title and show the plot
plt.title("Correlation Heatmap of Titanic Dataset (All Data Types)")
plt.show()
```

```python
# Histogram
columns_to_plot = ['age', 'fare', 'sex', 'class']  # You can change this list

# Loop through selected columns
for col in columns_to_plot:
    plt.figure(figsize=(4, 4))
    
    # Numeric columns → histogram
    if pd.api.types.is_numeric_dtype(titanic[col]):
        sns.histplot(titanic[col].dropna(), bins=30, kde=True, color='skyblue')
        plt.title(f'Histogram of {col}')
        plt.xlabel(col)
        plt.ylabel('Frequency')
    
    # Categorical columns → count plot
    else:
        sns.countplot(data=titanic, x=col, palette='Set2')
        plt.title(f'Count Plot of {col}')
        plt.xlabel(col)
        plt.ylabel('Count')
        plt.xticks(rotation=45)
    
    plt.tight_layout()
    plt.show()
```

```python
# Create a scatterplot
sns.scatterplot(data=titanic, x='age', y='fare', hue='sex', style='class')

# Add titles and labels
plt.title('Scatterplot of Fare vs Age')
plt.xlabel('Age')
plt.ylabel('Fare')

# Show the plot
plt.show()
```
